```python

from scipy.stats import norm

# Meerkat lifespans (years)
mu = 13.1
sigma = 1.5

# P(X < 10.1)
print("Less than 10.1:", norm.cdf(10.1, mu, sigma)*100)

# P(X > 14.6)
print("More than 14.6:", norm.sf(14.6, mu, sigma)*100)

# P(11.6 < X < 16.1)
prob = norm.cdf(16.1, mu, sigma) - norm.cdf(11.6, mu, sigma)
print("Between 11.6 and 16.1:", prob*100)
```

```python

# What is the probability a meerkat lives less than 15 years?
# Steps
# 1 - Calculate Z Score
# 2 - use norm.cdf or norm.sf to get probability of that z-scaore
x = 15
z = (x - mu) / sigma

print("Z-score:", z)
print("P(X < 15):", norm.cdf(z))
print("P(X > 15):", norm.sf(z))

# We could also calculate directly:
print("P(X < 15):", norm.cdf(15, mu, sigma))
```

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(7, 19, 500) # generate our x data (500 equally spaced values between 7 and 19)
y = norm.pdf(x, mu, sigma) # generate our y data - values for each x that follow the probability density of a normal curve based on the given population mean and standard deviation

plt.plot(x, y, label="Normal distribution") # add the curve data
plt.axvline(mu, linestyle="--", label="Mean = 13.1") # add a line for mean
plt.axvline(15, linestyle=":", label="15 years") # add a line for 15 years old

# Shade P(X < 15)
plt.fill_between(x, y, where=(x <= 15), alpha=0.3) # add filling for all x values < 15

plt.xlabel("Meerkat lifespan (years)")
plt.ylabel("Probability density")
plt.title("Probability of Living Less Than 15 Years")
plt.legend()
plt.show()
```

```python

# What lifespan corresponds to the 95th percentile?
lifespan = norm.ppf(0.95, mu, sigma)

print("95th percentile:", lifespan)
```
